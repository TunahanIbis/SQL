```python
import sqlite3

# Connect to the SQLite database
# This connection also creates a DB file in the current directory
connection = sqlite3.connect('MIRAMAX_movie_production.db')

# To run SQL commands, create a cursor object with a connection to the DB
# We will use this cursor frequently to update and use the DB
cursor = connection.cursor()

# Create PRODUCER table
# Attributes: PRODUCER = (p_id, p_name, telephone)
cursor.execute('''
CREATE TABLE IF NOT EXISTS PRODUCER (
    p_id INTEGER PRIMARY KEY,
    p_name TEXT NOT NULL,
    telephone TEXT
)
''')

# Create ACTOR table
# Attributes: ACTOR = (a_id, a_name, sex)
cursor.execute('''
CREATE TABLE IF NOT EXISTS ACTOR (
    a_id INTEGER PRIMARY KEY,
    a_name TEXT NOT NULL,
    sex TEXT
)
''')

# Create DIRECTOR table
# Attributes: DIRECTOR = (d_id, d_name, age)
cursor.execute('''
CREATE TABLE IF NOT EXISTS DIRECTOR (
    d_id INTEGER PRIMARY KEY,
    d_name TEXT NOT NULL,
    age INTEGER
)
''')

# Create THEATER table
# Attributes (from relational schema): THEATER = (t_id, t_name, city, country)
cursor.execute('''
CREATE TABLE IF NOT EXISTS THEATER (
    t_id INTEGER PRIMARY KEY,
    t_name TEXT NOT NULL,
    city TEXT NOT NULL,
    country TEXT NOT NULL
)
''')

# Create MOVIE table
# Attributes: MOVIE = (m_id, title, genre)
cursor.execute('''
CREATE TABLE IF NOT EXISTS MOVIE (
    m_id INTEGER PRIMARY KEY,
    title TEXT NOT NULL,
    genre TEXT
)
''')

# Create produces table
# Attributes: produces = (p_id, m_id)
# References PRODUCER and MOVIE tables
cursor.execute('''
CREATE TABLE IF NOT EXISTS produces (
    p_id INTEGER,
    m_id INTEGER,
    PRIMARY KEY (p_id, m_id),
    FOREIGN KEY (p_id) REFERENCES PRODUCER(p_id),
    FOREIGN KEY (m_id) REFERENCES MOVIE(m_id)
)
''')

# Create acts_in table
# Attributes: acts_in = (a_id, m_id)
# References ACTOR and MOVIE tables
cursor.execute('''
CREATE TABLE IF NOT EXISTS acts_in (
    a_id INTEGER,
    m_id INTEGER,
    PRIMARY KEY (a_id, m_id),
    FOREIGN KEY (a_id) REFERENCES ACTOR(a_id),
    FOREIGN KEY (m_id) REFERENCES MOVIE(m_id)
)
''')

# Create directs table
# Attributes: directs = (d_id, m_id)
# References DIRECTOR and MOVIE tables
cursor.execute('''
CREATE TABLE IF NOT EXISTS directs (
    d_id INTEGER,
    m_id INTEGER,
    PRIMARY KEY (d_id, m_id),
    FOREIGN KEY (d_id) REFERENCES DIRECTOR(d_id),
    FOREIGN KEY (m_id) REFERENCES MOVIE(m_id)
)
''')

# Create releases_in table
# Attributes: releases_in = (m_id, t_id)
# References MOVIE and THEATER tables
cursor.execute('''
CREATE TABLE IF NOT EXISTS releases_in (
    m_id INTEGER,
    t_id INTEGER,
    PRIMARY KEY (m_id, t_id),
    FOREIGN KEY (m_id) REFERENCES MOVIE(m_id),
    FOREIGN KEY (t_id) REFERENCES THEATER(t_id)
)
''')

# INSERTIONS
# Insert at least 5 different rows of data to the tables.
# Also, insert some multivalue data to the relationship tables because they are all M:N (many-to-many)

# Insert data into PRODUCER table
cursor.execute("INSERT INTO PRODUCER (p_name, telephone) VALUES ('Harvey Weinstein', '123-456-789')")
cursor.execute("INSERT INTO PRODUCER (p_name, telephone) VALUES ('Bob Weinstein', '987-654-321')")
cursor.execute("INSERT INTO PRODUCER (p_name, telephone) VALUES ('Quentin Tarantino', '555-111-223')")
cursor.execute("INSERT INTO PRODUCER (p_name, telephone) VALUES ('Lawrence Bender', '555-111-224')")
cursor.execute("INSERT INTO PRODUCER (p_name, telephone) VALUES ('David L. Wolper', '555-111-225')")

# Insert data into ACTOR table
cursor.execute("INSERT INTO ACTOR (a_name, sex) VALUES ('John Travolta', 'Male')")
cursor.execute("INSERT INTO ACTOR (a_name, sex) VALUES ('Uma Thurman', 'Female')")
cursor.execute("INSERT INTO ACTOR (a_name, sex) VALUES ('Samuel L. Jackson', 'Male')")
cursor.execute("INSERT INTO ACTOR (a_name, sex) VALUES ('Gwyneth Paltrow', 'Female')")
cursor.execute("INSERT INTO ACTOR (a_name, sex) VALUES ('Harvey Keitel', 'Male')")

# Insert data into DIRECTOR table
cursor.execute("INSERT INTO DIRECTOR (d_name, age) VALUES ('Quentin Tarantino', 60)")
cursor.execute("INSERT INTO DIRECTOR (d_name, age) VALUES ('Ang Lee', 70)")
cursor.execute("INSERT INTO DIRECTOR (d_name, age) VALUES ('Robert Rodriguez', 55)")
cursor.execute("INSERT INTO DIRECTOR (d_name, age) VALUES ('Martin Scorsese', 81)")
cursor.execute("INSERT INTO DIRECTOR (d_name, age) VALUES ('Tim Burton', 65)")

# Insert data into THEATER table
cursor.execute("INSERT INTO THEATER (t_name, city, country) VALUES ('Miramax Theater', 'Los Angeles', 'USA')")
cursor.execute("INSERT INTO THEATER (t_name, city, country) VALUES ('Sunset Cinema', 'Los Angeles', 'USA')")
cursor.execute("INSERT INTO THEATER (t_name, city, country) VALUES ('Miramax Cinema', 'New York', 'USA')")
cursor.execute("INSERT INTO THEATER (t_name, city, country) VALUES ('West End Theater', 'London', 'UK')")
cursor.execute("INSERT INTO THEATER (t_name, city, country) VALUES ('Le Grand Rex', 'Paris', 'France')")

# Insert data into MOVIE table
cursor.execute("INSERT INTO MOVIE (title, genre) VALUES ('Pulp Fiction', 'Crime')")
cursor.execute("INSERT INTO MOVIE (title, genre) VALUES ('Kill Bill: Vol. 1', 'Action')")
cursor.execute("INSERT INTO MOVIE (title, genre) VALUES ('Good Will Hunting', 'Drama')")
cursor.execute("INSERT INTO MOVIE (title, genre) VALUES ('Shakespeare in Love', 'Romance')")
cursor.execute("INSERT INTO MOVIE (title, genre) VALUES ('The English Patient', 'Romance')")

# Insert data into produces table
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (1, 1)")
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (1, 2)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (2, 3)")
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (2, 4)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (3, 3)")
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (4, 4)")
cursor.execute("INSERT INTO produces (p_id, m_id) VALUES (5, 5)")

# Insert data into acts_in table
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (1, 2)")
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (2, 1)")
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (3, 1)")
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (2, 2)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (3, 3)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (4, 4)") 
cursor.execute("INSERT INTO acts_in (a_id, m_id) VALUES (5, 5)")  

# Insert data into directs table
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (1, 1)")
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (1, 2)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (2, 2)")
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (3, 3)")
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (4, 4)")
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (4, 5)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO directs (d_id, m_id) VALUES (5, 5)")

# Insert data into releases_in table
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (1, 1)")
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (1, 2)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (2, 2)")
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (3, 3)")
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (3, 2)") # Multivalue data (the first primary key is used earlier)
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (4, 4)")
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (5, 4)")
cursor.execute("INSERT INTO releases_in (m_id, t_id) VALUES (5, 5)") # Multivalue data (the first primary key is used earlier)

# Commit the changes to the database
connection.commit()

# Close the cursor and connection
cursor.close()
connection.close()

# Notify the user
print("Data inserted into the database successfully.")
```

```python
import sqlite3

# Specify the table options to let user make choices only with integers
TABLE_OPTIONS = {
    "1": "MOVIE",
    "2": "PRODUCER",
    "3": "ACTOR",
    "4": "DIRECTOR",
    "5": "THEATER",
    "6": "produces",
    "7": "acts_in",
    "8": "directs",
    "9": "releases_in"
}

# Connect to the SQLite database
def connect_db():
    return sqlite3.connect('MIRAMAX_movie_production.db')

# Display table options
def select_table():
    print("\n--- Select a Table ---")
    for key, value in TABLE_OPTIONS.items():
        print(f"{key}. {value}")
    choice = input("Enter your table choice: ").strip() # Get user choice for table
    return TABLE_OPTIONS.get(choice)

# DISPLAY FUNCTION
# Display the records from the table selected by user
def display_table_records():
    table_name = select_table() # Get table choice from user
    if not table_name:
        print("Invalid choice. Please try again.")
        return

    connection = connect_db()
    cursor = connection.cursor()
    try:
        cursor.execute(f"SELECT * FROM {table_name}") # Select the desired table with SQL command
        rows = cursor.fetchall()
        print(f"\n--- {table_name} Records ---")
        for row in rows:
            print(row)
    except sqlite3.Error as e:
        print(f"Error: {e}")
    finally:
        cursor.close()
        connection.close()

# ADD RECORD FUNCTION
# Add records to a table selected by user
def add_record():
    table_name = select_table() # Get table choice from user
    if not table_name:
        print("Invalid choice. Please try again.")
        return

    connection = connect_db()
    cursor = connection.cursor()

    try:
        cursor.execute(f"PRAGMA table_info({table_name})") # Make operations on the selected table
        columns = cursor.fetchall()
        values = []
        print(f"\nAdding a record to {table_name}:") 
        for column in columns:
            col_name = column[1]
            col_value = input(f"Enter value for {col_name}: ") # Get the value to add from user
            values.append(col_value)

        placeholders = ", ".join(["?" for _ in values])
        query = f"INSERT INTO {table_name} VALUES ({placeholders})" # Insert the data with SQL command
        cursor.execute(query, values)
        connection.commit()
        print("Record added successfully!") # Notify the user
    except sqlite3.Error as e:
        print(f"Error: {e}")
    finally:
        cursor.close()
        connection.close()

# DELETE RECORD FUNCTION
# Delete records from a table selected by user
def delete_record():
    table_name = select_table() # Get table choice from user
    if not table_name:
        print("Invalid choice. Please try again")
        return

    connection = connect_db()
    cursor = connection.cursor()

    try:
        cursor.execute(f"PRAGMA table_info({table_name})") # Make operations on the selected table
        primary_key_column = cursor.fetchall()[0][1]
        print(f"\nDeleting a record from {table_name}:") # Delete record from the selected table
        primary_key_value = input(f"Enter the value of {primary_key_column} to delete: ").strip() # Get the value to be deleted from the user

        query = f"DELETE FROM {table_name} WHERE {primary_key_column} = ?" # Delete the data with SQL command
        cursor.execute(query, (primary_key_value,))
        connection.commit()
        print(f"Record with {primary_key_column}={primary_key_value} deleted successfully!") # Notify the user
    except sqlite3.Error as e:
        print(f"Error: {e}")
    finally:
        cursor.close()
        connection.close()

# UPDATE RECORD FUNCTION
# Update records in a selected table
def update_record():
    table_name = select_table() # Get table choice from user
    if not table_name:
        print("Invalid choice. Please try again")
        return

    connection = connect_db()
    cursor = connection.cursor()

    try:
        cursor.execute(f"PRAGMA table_info({table_name})") # Make operations on the selected table
        columns = cursor.fetchall()

        if not columns:
            print("No columns found in the table.")
            return

        # Get the primary key column to identify the record
        primary_key_column = columns[0][1]
        print(f"\nUpdating a record in {table_name}:") # Update the selected table
        primary_key_value = input(f"Enter the value of {primary_key_column} to update: ").strip() # Get the value to be updated from the user

        # Fetch the record with the provided primary key
        cursor.execute(f"SELECT * FROM {table_name} WHERE {primary_key_column} = ?", (primary_key_value,))
        record = cursor.fetchone()

        if not record:
            print(f"No record found with {primary_key_column}={primary_key_value}.")
            return

        print(f"Record found: {record}")
        updated_values = []

        # Get new values for each columns from the user
        for idx, column in enumerate(columns):
            col_name = column[1]
            col_value = input(f"Enter new value for {col_name} (leave blank to keep current value): ").strip() # Get the new value from user
            
            if not col_value:  # If the input is blank, keep the current value
                col_value = record[idx]
            
            updated_values.append(col_value)

        # Create the UPDATE query
        set_clause = ", ".join([f"{columns[idx][1]} = ?" for idx in range(len(updated_values))])
        query = f"UPDATE {table_name} SET {set_clause} WHERE {primary_key_column} = ?" # Update the table with SQL command

        # Execute the update
        cursor.execute(query, (*updated_values, primary_key_value))
        connection.commit()
        print("Record updated successfully!") # Notify the user

    except sqlite3.Error as e:
        print(f"Error: {e}")
    finally:
        cursor.close()
        connection.close()

# JOIN FUNCTION
# Execute join operation to merge all the data
def execute_join():
    connection = connect_db()
    cursor = connection.cursor()

    try:
        print("\nExecuting JOIN operation to display movies with producers, actors, directors, and theaters:")

        # Join the tables with GROUP and CONCAT methods to merge the unique data
        # We use these methods in order to print multivalue data correctly
        cursor.execute('''
            SELECT 
                m.title AS Movie_Title,
                m.genre AS Genre,
                (SELECT GROUP_CONCAT(p_name, ', ') 
                 FROM (SELECT DISTINCT p.p_name FROM PRODUCER p 
                       JOIN produces pr ON p.p_id = pr.p_id 
                       WHERE pr.m_id = m.m_id)) AS Producers,
                (SELECT GROUP_CONCAT(a_name, ', ') 
                 FROM (SELECT DISTINCT a.a_name FROM ACTOR a 
                       JOIN acts_in ai ON a.a_id = ai.a_id 
                       WHERE ai.m_id = m.m_id)) AS Actors,
                (SELECT GROUP_CONCAT(d_name, ', ') 
                 FROM (SELECT DISTINCT d.d_name FROM DIRECTOR d 
                       JOIN directs di ON d.d_id = di.d_id 
                       WHERE di.m_id = m.m_id)) AS Directors,
                (SELECT GROUP_CONCAT(t_name, ', ') 
                 FROM (SELECT DISTINCT t.t_name FROM THEATER t 
                       JOIN releases_in ri ON t.t_id = ri.t_id 
                       WHERE ri.m_id = m.m_id)) AS Theaters
            FROM MOVIE m
        ''')

        rows = cursor.fetchall()

        print("\n--- Movies with Producers, Actors, Directors, and Theaters ---")
        for row in rows:
            print(row)

    except sqlite3.Error as e:
        print(f"Error: {e}")
    finally:
        cursor.close()
        connection.close()

# Main menu function
def main_menu():
    while True:
        # Display the menu to the user
        print("\n--- Miramax Movie Production Database ---")
        print("1. Add a record")
        print("2. Delete a record")
        print("3. Update a record")
        print("4. Display records from a table")
        print("5. Execute JOIN operation")
        print("6. Exit")

        choice = input("Choose your operation: ")
        if choice == '1':
            add_record()
        elif choice == '2':
            delete_record()
        elif choice == '3':
            update_record()
        elif choice == '4':
            display_table_records()
        elif choice == '5':
            execute_join()
        elif choice == '6':
            print("Exiting the application. Come back later!")
            break
        else:
            print("Invalid choice. Please try again.")

# Run the application
if __name__ == "__main__":
    main_menu()
```
